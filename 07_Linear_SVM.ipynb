# Support Vector Machines

- In their basic version, they are *linear classifiers*
- Instead of minimizing a classification loss, they maximize the $\textbf{Margin}$
- The $\textbf{Margin}$ is the distance between the decision boundary (a hyperplane) and the closest inputs (support vectors)

### Max-margin classification
Let $\left\{\mathbf{x} \in \mathbb{R}^{m}: \mathbf{w}^{\top} \mathbf{x}+w_0 = 0\right\}$ be the decision boundary

Define two hyperplanes (*positive* and *negative margin hyperplane*) that are parallel to and equidistant from the decision boundary

$$\begin{array}{l}
\left\{\mathbf{x}_{pos} \in \mathbb{R}^{m}: \mathbf{w}^{\top} \mathbf{x}_{pos}+w_0 = 1\right\} \\
\left\{\mathbf{x}_{neg} \in \mathbb{R}^{m}: \mathbf{w}^{\top} \mathbf{x}_{neg}+w_0 = -1\right\}
\end{array}$$

For example, if $\mathbf{w}=\left[\begin{array}{c}
  2 \\
  1
\end{array}\right]$ and $w_0=-6$, the three hyperplanes are described by the following equations:

$$2x_1 + x_2 - 6 = 1$$ (positive margin hyperplane)
$$2x_1 + x_2 - 6 = 0$$ (decision boundary)
$$2x_1 + x_2 - 6 = -1$$ (negative margin hyperplane)

```python
import numpy as np
import matplotlib.pyplot as plt

w = np.array([-6,2,1])

x1 = np.linspace(-4,6,100)
x2 = (-w[1]*x1 - w[0])/w[2]
x2_pos = (-w[1]*x1 - (w[0]-1))/w[2]
x2_neg = (-w[1]*x1 - (w[0]+1))/w[2]

plt.figure(figsize=(15, 8))
plt.axvline(0.0, color='k')
plt.axhline(0.0, color='k')
plt.plot(x1,x2, label='$2x_1 + x_2 - 6 = 0$')
plt.plot(x1,x2_pos, '--', label='$2x_1 + x_2 - 6 = 1$')
plt.plot(x1,x2_neg, '--', label='$2x_1 + x_2 - 6 = -1$')
plt.xlabel('x1')
plt.ylabel('x2')
plt.legend()
plt.xlim(-2, 6)
plt.ylim(-2, 10)
plt.show()
```

Multiplying all weights by a constant $k>0$: $\mathbf{w}' = k\mathbf{w}$, $w_0'=kw_0$

```python
k = 0.1
w = np.array([-6,2,1]) * k

x1 = np.linspace(-4,6,100)
x2 = (-w[1]*x1 - w[0])/w[2]
x2_pos = (-w[1]*x1 - (w[0]-1))/w[2]
x2_neg = (-w[1]*x1 - (w[0]+1))/w[2]

plt.figure(figsize=(12, 6))
plt.axvline(0.0, color='k')
plt.axhline(0.0, color='k')
plt.plot(x1,x2, label='$2x_1 + x_2 - 6 = 0$')
plt.plot(x1,x2_pos, '--', label='$2x_1 + x_2 - 6 = 1$')
plt.plot(x1,x2_neg, '--', label='$2x_1 + x_2 - 6 = -1$')
plt.xlabel('x1')
plt.ylabel('x2')
plt.legend()
plt.xlim(-2, 6)
plt.ylim(-2, 10)
plt.show()

print(np.linalg.norm(w))
```

The distance between the two hyperplanes is

$$\frac{\mathbf{w}^{T}\left(\mathbf{x}_{\text {pos }}-\mathbf{x}_{\text {neg }}\right)}{\|\mathbf{w}\|^2}=\frac{2}{\|\mathbf{w}\|}$$


To maximize margin, maximize $\frac{2}{\|\mathbf{w}\|}$ or, equivalently, minimize $\frac{1}{2}\|\mathbf{w}\|^{2}$

### Correct classification
Let $\mathbf{x}^{(i)}$ be the $i$-th training sample and $y^{(i)} \in \{-1,1\}$ the corresponding label

Minimize $\frac{1}{2}\|\mathbf{w}\|^{2}$ subject to the constraint that all examples are classified correctly and lie outside of the margin:

$$\begin{array}{c}
\mathbf{w}^{T} \mathbf{x}^{(i)} +w_0\geq 1 \text { if } y^{(i)}=1 \\
\mathbf{w}^{T} \mathbf{x}^{(i)} +w_0\leq-1 \text { if } y^{(i)}=-1 \\
\forall i \in \{1 \ldots N\}
\end{array}$$

More compactly:

$$y^{(i)}(\mathbf{w}^{T} \mathbf{x}^{(i)}+w_0) \geq 1 \quad \forall_{i}$$

The (primal) SVM optimization problem:

$$\begin{array}{cl}
\min _{\mathbf{w} \in \mathbb{R}^{D}, w_0\in\mathbb{R}} & \frac{1}{2}\|\mathbf{w}\|^{2} \\
\text { s.t. } & y^{(i)} (\mathbf{w}^{\top} \mathbf{x}^{(i)}+w_0) \geq 1 \quad i=1, \ldots, N
\end{array}$$

The corresponding *dual optimization problem* is

$$\begin{array}{cl}
\max_{\boldsymbol\alpha}\quad&
\sum_{n=1}^N\alpha_n
-\frac12\sum_{n=1}^N\sum_{m=1}^N
\alpha_n\alpha_m t_nt_m\mathbf{x}_n^\top\mathbf{x}_m\\
\text{subject to}\quad&\alpha_n\geq0,\quad n=1,\ldots,N,\\
&\sum_{n=1}^N\alpha_n t_n=0
\end{array}$$

The optimal solution (margin-maximizing weights) can be computed as:

$$\mathbf{w}_{*}=\sum_{i \in I} \alpha^{(i)} y^{(i)} \mathbf{x}^{(i)}$$

Where $I$ is the set of training samples for which $y^{(i)}\mathbf{w}_{*}^{\top} \mathbf{x}^{(i)}=1$.

- These $\mathbf{x}^{(i)}$ are called $\textbf{support vectors}$ and lie exactly on the margin planes

- If we remove all samples from the training set except the support vectors, we obtain the same classifier

## SVM in sklearn
Linear SVM for classification are implemented in the LinearSVC class ([doc](https://scikit-learn.org/stable/modules/generated/sklearn.svm.LinearSVC.html))

```python
import pandas as pd
s = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
df = pd.read_csv(s, header=None, names=['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width', 'Species'])
df = df.iloc[1:]
df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import ListedColormap
from sklearn.preprocessing import StandardScaler

def plot_decision_regions(X, y, classifier, resolution=0.02):

    # setup marker generator and color map
    markers = ('s', 'x', 'o', '^', 'v')
    colors = ('red', 'blue', 'lightgreen', 'gray', 'cyan')
    cmap = ListedColormap(colors[:len(np.unique(y))])

    # plot the decision surface
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(np.arange(x1_min, x1_max, resolution),
                           np.arange(x2_min, x2_max, resolution))
    Z = classifier.predict(np.array([xx1.ravel(), xx2.ravel()]).T)
    Z = Z.reshape(xx1.shape)
    plt.contourf(xx1, xx2, Z, alpha=0.3, cmap=cmap)
    plt.xlim(xx1.min(), xx1.max())
    plt.ylim(xx2.min(), xx2.max())

    # plot class examples
    for idx, cl in enumerate(np.unique(y)):
        edge_kwargs = {} if markers[idx] == 'x' else {'edgecolor': 'black'}

        plt.scatter(
            x=X[y == cl, 0],
            y=X[y == cl, 1],
            alpha=0.8,
            c=colors[idx],
            marker=markers[idx],
            label=cl,
            **edge_kwargs
        )

    # Draw the binary classifier's decision boundary: f(x) = 0
    grid = np.c_[xx1.ravel(), xx2.ravel()]
    scores = classifier.decision_function(grid).reshape(xx1.shape)

    plt.contour(
        xx1, xx2, scores,
        levels=[0],
        colors="black",
        linewidths=1.5
    )

# select setosa and versicolor
y = df.iloc[0:100, 4].values
y_iris = np.where(y == 'Iris-setosa', -1, 1)

# extract sepal length and petal length
X_iris = df.iloc[0:100, [0, 2]].values

X = np.copy(X_iris)
y = np.copy(y_iris)

#Standardize data
scaler = StandardScaler()
X = scaler.fit_transform(X)

# plot data
plt.scatter(X[:49, 0], X[:49, 1],
            color='red', marker='o', label='setosa')
plt.scatter(X[50:99, 0], X[50:99, 1],
            color='blue', marker='x', label='versicolor')

plt.xlabel('sepal length [cm]')
plt.ylabel('petal length [cm]')
plt.legend(loc='upper left')

plt.show()
```

Let's train a **hard-margin** SVM

```python
from sklearn.svm import LinearSVC

svm = LinearSVC(random_state=123, loss="hinge", C=1e6)  # Very large C approximates "hard margin"
svm.fit(X, y)

plot_decision_regions(X, y, classifier=svm)
plt.xlabel('petal length [standardized]')
plt.ylabel('petal width [standardized]')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

Compare this with the solution found by the Perceptron algorithm

```python
from sklearn.linear_model import Perceptron

ppn = Perceptron(eta0=0.1, random_state=123)
ppn.fit(X, y)

plot_decision_regions(X, y, classifier=ppn)
plt.xlabel('petal length [standardized]')
plt.ylabel('petal width [standardized]')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

**LinearSVC** uses a solver optimized for linear classification, and is especially faster on large datasets.

We will now switch to the more general, although less efficient **SVC** class for visualization purposes ([doc](https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVC.html#sklearn.svm.SVC))

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

def plot_decision_regions(X, y, classifier, resolution=0.02,
                          support_vectors=None):
    X, y = np.asarray(X), np.asarray(y)
    classes = np.unique(y)
    if len(classes) != 2:
        raise ValueError("This function requires binary classification.")

    colors = ("red", "blue")
    markers = ("s", "x")
    cmap = ListedColormap(colors)

    # Grid covering the data
    x1_min, x1_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    x2_min, x2_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx1, xx2 = np.meshgrid(
        np.arange(x1_min, x1_max, resolution),
        np.arange(x2_min, x2_max, resolution)
    )
    grid = np.c_[xx1.ravel(), xx2.ravel()]

    # Colored class regions
    predictions = classifier.predict(grid)
    Z = np.searchsorted(classes, predictions).reshape(xx1.shape)
    plt.contourf(xx1, xx2, Z, levels=[-0.5, 0.5, 1.5],
                 alpha=0.2, cmap=cmap)

    # Decision boundary and margins: f(x) = 0 and f(x) = ±1
    scores = classifier.decision_function(grid).reshape(xx1.shape)
    plt.contour(
        xx1, xx2, scores,
        levels=[-1, 0, 1],
        colors="black",
        linestyles=["--", "-", "--"],
        linewidths=[1, 1.5, 1]
    )

    # Samples
    for idx, cl in enumerate(classes):
        edge_kwargs = {} if markers[idx] == "x" else {"edgecolor": "black"}
        plt.scatter(
            X[y == cl, 0], X[y == cl, 1],
            color=colors[idx], marker=markers[idx],
            alpha=0.8, label=cl, zorder=3, **edge_kwargs
        )

    # Highlight supplied samples with rings
    if support_vectors is not None:
        sv = np.asarray(support_vectors).reshape(-1, 2)
        if len(sv):
            plt.scatter(
                sv[:, 0], sv[:, 1],
                s=160, facecolors="none", edgecolors="black",
                linewidths=1.5, label="Support vectors", zorder=4
            )

    plt.xlim(x1_min, x1_max)
    plt.ylim(x2_min, x2_max)
    plt.gca().set_aspect("equal", adjustable="box")
```

```python
from sklearn.svm import SVC

svm = SVC(kernel="linear", C=1e6)  # linear classification and (approximate) hard margin
svm.fit(X, y)

# Extract support vectors and their indices in X
support_vectors = svm.support_vectors_
support_indices = svm.support_

print("Found %d support vectors" % len(svm.support_vectors_))

plot_decision_regions(
    X, y, classifier=svm,
    support_vectors=support_vectors
)
plt.xlabel('petal length [standardized]')
plt.ylabel('petal width [standardized]')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

We can see the dual variables associated with the support vectors (other samples have dual variables equal to zero)

```python
svm.dual_coef_
```

Let's try to re-train the classifier using only the support vectors!

```python
X_tiny = svm.support_vectors_
y_tiny = y[svm.support_]

svm_tiny = SVC(kernel="linear", C=1e6)  # linear classification and (approximate) hard margin
svm_tiny.fit(X_tiny, y_tiny)

plot_decision_regions(
    X_tiny, y_tiny, classifier=svm_tiny,
    support_vectors=svm_tiny.support_vectors_
)
plt.xlabel('petal length [standardized]')
plt.ylabel('petal width [standardized]')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

## Soft-margin SVM

Hard-margin SVM has no solution for data that are not linearly separable (the optimization problem becomes infeasible)

Solution:

- __Slack variables__: constraints can be relaxed, but are penalized in the objective function

$$\begin{array}{c}
\mathbf{w}^{T} \mathbf{x}^{(i)} + w_0 \geq 1-\xi^{(i)} \text { if } y^{(i)}=1 \\
\mathbf{w}^{T} \mathbf{x}^{(i)} + w_0 \leq-1+\xi^{(i)} \text { if } y^{(i)}=-1 \\
\forall i \in \{1 \ldots N\}
\end{array}$$

More concisely:

$$\begin{array}{l}
y^{(i)}(\mathbf{w}^{T}\mathbf{x}^{(i)}+w_0) \geqslant 1-\xi^{(i)} \\
\xi^{(i)} \geqslant 0
\end{array}$$

The new *primal* optimization problem is:

$$\begin{array}{cl}
\min _{\mathbf{w}, \xi} & \frac{1}{2}\|\mathbf{w}\|^{2}+C \sum_{i=1}^{N} \xi_{i} \\
\text { s.t. } & y^{(i)}(\mathbf{w}^{T}\mathbf{x}^{(i)}+w_0) \geqslant 1-\xi^{(i)} \\
& \xi^{(i)} \geqslant 0
\end{array}$$

The *hyperparameter* $C>0$ controls how much missclassified or margin-violating samples are penalized.

- Large $C$ will prioritize class separation and will tend to produce a smaller margin

- Small $C$ will tend to produce a larger margin but more violations

### Soft-margin SVM in scikit-learn

```python
from sklearn.datasets import make_moons

X, y = make_moons(noise=0.3, random_state=0)

# plot data
cm_bright = ListedColormap(['red', 'blue'])
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=cm_bright, edgecolors='k')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.show()
```

Using **LinearSVC**:

```python
from sklearn.svm import LinearSVC

C = [0.05, 1000]

plt.figure(figsize=(13,5))
for i,c in enumerate(C):
    plt.subplot(1,2,i+1)
    svm = LinearSVC(C=c, max_iter=100000, random_state=123)
    svm.fit(X, y)

    plot_decision_regions(X, y, classifier=svm)
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.legend(loc='upper left')
    plt.tight_layout()
    plt.title('C=' + str(c))
plt.show()
```

Using **SVC** (all margin-violating samples are support vectors too)

```python
from sklearn.svm import SVC

C = [0.05, 1000]

plt.figure(figsize=(13, 5))

for i, c in enumerate(C):
    plt.subplot(1, 2, i + 1)

    svm = SVC(kernel="linear", C=c)
    svm.fit(X, y)

    plot_decision_regions(
        X, y, classifier=svm,
        support_vectors=svm.support_vectors_
    )
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.legend(loc="upper left")
    plt.title(f"C = {c}")

plt.tight_layout()
plt.show()
```
