## Dot Product

$$
\mathbf{a}^{\top} \mathbf{b} = 
\begin{bmatrix}
a_{1} & a_{2} & a_{3} & \cdots & a_{D}
\end{bmatrix}
\times
\begin{bmatrix}
b_{1} \\
b_{2} \\
b_{3} \\
\vdots \\
b_{D}
\end{bmatrix}
= a_{1}b_{1} + a_{2}b_{2} + a_{3}b_{3} + \cdots + a_{D}b_{D} = \sum_{i=1}^{D} a_{i}b_{i}
$$

## Matrix product

Given two matrixes $\mathbf{A} \in \mathbb{R}^{m\times n}$ and $\mathbf{B} \in \mathbb{R}^{n\times p}$ their product is the matrix $\mathbf{C} \in \mathbb{R}^{m\times p}$ of elements:

$c_{i j}= \mathbf{a}_i^T\mathbf{b}_j = \sum_{k=1}^{n} a_{i k} b_{k j}$

or, the matrix of the internal products of the rows of $\mathbf{A}$ and of the columns of $\mathbf{B}$ (product row-column)

$$
\mathbf{C} = 
\begin{bmatrix}
- & \mathbf{a}_{1}^{\top} & - \\
- & \mathbf{a}_{2}^{\top} & - \\
& \vdots & \\
- & \mathbf{a}_{m}^{\top} & -
\end{bmatrix}
\begin{bmatrix}
\mid & \mid & & \mid \\
\mathbf{b}_{1} & \mathbf{b}_{2} & \cdots & \mathbf{b}_{p} \\
\mid & \mid & & \mid
\end{bmatrix}
=
\begin{bmatrix}
\mathbf{a}_{1}^{\top} \mathbf{b}_{1} & \mathbf{a}_{1}^{\top} \mathbf{b}_{2} & \cdots & \mathbf{a}_{1}^{\top} \mathbf{b}_{p} \\
\mathbf{a}_{2}^{\top} \mathbf{b}_{1} & \mathbf{a}_{2}^{\top} \mathbf{b}_{2} & \cdots & \mathbf{a}_{2}^{\top} \mathbf{b}_{p} \\
\vdots & \vdots & \ddots & \vdots \\
\mathbf{a}_{m}^{\top} \mathbf{b}_{1} & \mathbf{a}_{m}^{\top} \mathbf{b}_{2} & \cdots & \mathbf{a}_{m}^{\top} \mathbf{b}_{p}
\end{bmatrix}
$$

## Inverse Matrix

If the determinant of a square matrix $\mathbf{M}$ is different from zero, we can calculate the reverse matrix $\mathbf{M}^{-1}$.

$\mathbf{M}\mathbf{M}^{-1} = \mathbf{M}^{-1}\mathbf{M} = \mathbf{I}$

where $\mathbf{I}$ is the identity matrix:

$$
\mathbf{I} = 
\begin{bmatrix}
1 & 0 & \cdots & 0 \\
0 & 1 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & 1
\end{bmatrix}
$$

```python
M = np.array([[5,12,6],[20,2,8],[3,87,12]])
Minv = np.linalg.inv(M)
```

## Moore-Penrose Pseudo-inverse

If the $\mathbf{M}$ matrix is not square we can calculate the pseudo-inverse matrix:

$\mathbf{M}^{\dagger}=\left(\mathbf{M}^{\top} \mathbf{M}\right)^{-1} \mathbf{M}^{\top}$

In this case, the following shall apply:

$\mathbf{M}^{\dagger} \mathbf{M}=\mathbf{I}$

```python
M = np.random.rand(3,2)
M
```

```python
pseudoM1 = np.linalg.pinv(M)
pseudoM2 = np.linalg.inv(M.T @ M) @ M.T
```

# Machine Learning in Python: The Scikit-learn package

`scikit-learn` is Python's standard library for classical Machine Learning. It provides an extensive ecosystem of algorithms for **supervised learning** (classification, regression) and **unsupervised learning** (clustering, dimensionality reduction), alongside tools for data preprocessing and model evaluation.

It is built directly on top of the Python scientific computing stack: **NumPy** (array operations), **SciPy** (linear algebra/optimization), and **Matplotlib** (visualization).

---

### The 3 Core Architectural Pillars

Regardless of the underlying mathematical model, `scikit-learn` enforces a unified, consistent API built on three abstraction patterns:

1. **Data Representation:**
   - **Feature Matrix ($\mathbf{X}$):** A 2D array of shape `(n_samples, n_features)` containing input measurements.
   - **Target Vector ($\mathbf{y}$):** A 1D array of shape `(n_samples,)` containing target labels or continuous values.

2. **Transformers (Preprocessing):**
   - Objects that clean, scale, or modify data features (e.g., feature scalers, encoders).
   - Core Methods: `.fit()` to learn parameters from training data, and `.transform()` to apply the transformation.

3. **(Meta)Estimators (Models):**
   - Objects that implement learning algorithms (classifiers, regressors, clusterers).
   - Core Methods: `.fit(X, y)` to train the model, and `.predict(X)` to generate predictions on new data.

### General Workflow Blueprint

Here is a template demonstrating how a supervised machine learning pipeline could be constructed in `scikit-learn`:

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import numpy as np

# Sample Data
X_train = np.array([[1.0, 20.0], [2.0, 40.0], [3.0, 60.0], [4.0, 80.0]])
y_train = np.array([0, 0, 1, 1])
X_test = np.array([[2.5, 50.0]])

# Transformer Lifecycle: Instantiate -> Fit -> Transform
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)  # Apply exact same parameters to test data!

# Predictor Lifecycle: Instantiate -> Fit -> Predict / Score
model = LogisticRegression(C=1.0, max_iter=100)  # Hyperparameters set in constructor
model.fit(X_train_scaled, y_train)
predictions = model.predict(X_test_scaled)
probabilities = model.predict_proba(X_test_scaled)

print(f"Scaled Test Instance: {X_test_scaled}")
print(f"Predicted Class: {predictions[0]}")
print(f"Class Probabilities: {probabilities}")
```

# Feature Engineering & Preprocessing Tasks

Feature Engineering converts raw, unformatted data into informative feature vectors $\{(\mathbf{x}_i, y_i)\}_{i=1}^n$ to maximize predictive performance.   

### Data Preprocessing Tasks

- **Data Cleaning**: Identifying and fixing missing values, correcting inconsistencies, and handling outliers.   

- **Data Transformation**: Rescaling continuous features, mapping categorical features, and extracting representations.   

- **Data Reduction**: Reducing dataset dimensionality via feature selection, extraction, or sampling.   

- **Data Discretization**: Partitioning continuous variables into discrete categories (binning).   

- **Data Formatting**: Converting raw types, timestamps, and structure constraints.   

## Handling Missing Data

Missing entries (represented as `NaN`, `NULL`, or empty placeholders) disrupt standard mathematical estimators and must be handled.   

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Sample dataset with missing values
raw_data = {
    'Age': [25, 30, np.nan, 45, 110],
    'Salary': [50000, np.nan, 60000, 80000, np.nan],
    'Department': ['HR', 'IT', 'IT', np.nan, 'HR']
}
df = pd.DataFrame(raw_data)
print(df)
```

### Methods:

- **Elimination**: Dropping entire rows (`axis=0`) or columns (`axis=1`) using Pandas `dropna()`:   

    - `how='all'`: Drop only if all attributes are missing.   
    - `thresh=k`: Retain rows containing at least $k$ non-null entries.   
    - `subset=['col1', 'col2']`: Check for missing values within specified columns.   
    
- **Imputation**: Estimating missing values via summary statistics:   
    - Mean / Median Imputation: Suited for numerical columns.   
    - Mode Imputation: Replaces missing entries with the most frequent value (ideal for categorical data). 

```python
# 1. Identification
print("Missing Count Per Column:")
print(df.isnull().sum())
```

```python
# 2. Elimination Example (Pandas)
df_dropped = df.dropna()
print(df_dropped)
```

```python
# 3. Imputation using Scikit-Learn SimpleImputer (median)
num_imputer = SimpleImputer(missing_values=np.nan, strategy='median')
df[['Age', 'Salary']] = num_imputer.fit_transform(df[['Age', 'Salary']])
```

```python
# 3. Imputation using Scikit-Learn SimpleImputer
cat_imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
df[['Department']] = cat_imputer.fit_transform(df[['Department']])
```

```python
print("\nImputed Dataframe:")
print(df)
```

## Handling Categorical Data

Categorical attributes contain discrete string or symbolic representations. They are split into Ordinal (naturally ordered) and Nominal (unordered) features.   

### Approaches

- **Ordinal Encoding**: Replaces categories with integer ranks preserving sequence order.   

    - *Pandas*: Direct dictionary mapping via `.map(dict)`.   
    - *Scikit-Learn*: OrdinalEncoder.   
    
- **One-Hot Encoding**: Creates dynamic binary indicator columns for each distinct categorical value. To prevent multicollinearity (dummy variable trap), drop one binary column.   
    
    - *Pandas*: `pd.get_dummies(df, drop_first=True)`.   
    - *Scikit-Learn*: `OneHotEncoder(drop='first', sparse_output=False)`. 

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

df_cat = pd.DataFrame({
    'Education': ['High School', 'Bachelor', 'Master', 'Bachelor'],  # Ordinal
    'City': ['London', 'Paris', 'London', 'Sallisaw']               # Nominal
})
print(df_cat)
```

```python
# --- Ordinal Encoding ---
# Strategy A: Pandas Map
education_order = {'High School': 0, 'Bachelor': 1, 'Master': 2}
df_cat['Education_Pandas'] = df_cat['Education'].map(education_order)
print(df_cat)
```

```python
# --- Ordinal Encoding ---
# Strategy B: Scikit-Learn OrdinalEncoder
ord_enc = OrdinalEncoder(categories=[['High School', 'Bachelor', 'Master']])
df_cat['Education_SKLearn'] = ord_enc.fit_transform(df_cat[['Education']])
print(df_cat)
```

```python
# --- Nominal Encoding (One-Hot) ---
# Strategy A: Pandas get_dummies
df_dummies = pd.get_dummies(df_cat['City'], dtype=int)
print(df_dummies)
```

```python
# --- Nominal Encoding (One-Hot) ---
# Strategy B: Scikit-Learn OneHotEncoder
ohe = OneHotEncoder(sparse_output=False, dtype=int)
city_ohe = ohe.fit_transform(df_cat[['City']])
city_ohe_df = pd.DataFrame(city_ohe, columns=ohe.get_feature_names_out(['City']))
print(city_ohe_df)
```

```python
print("Encoded DataFrame with One-Hot Columns:")
print(pd.concat([df_cat, city_ohe_df], axis=1))
```

## Data Discretization (Binning)

Data discretization converts continuous numerical attributes into discrete ordinal categories. Binning suppresses outlier impacts and simplifies non-linear relationship modeling.   

### Strategies

- **Uniform Width** `(strategy='uniform')`: Creates equal-interval bins.   
- **Quantile** `(strategy='quantile')`: Creates bins with an equal count of samples.   

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer

# Synthetic continuous feature
np.random.seed(42)
income = np.random.normal(loc=50000, scale=15000, size=(1000, 1))

# KBinsDiscretizer configuration
discretizer = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform')
income_binned = discretizer.fit_transform(income)

df_binned = pd.DataFrame({
    'Raw Income': income.ravel(),
    'Bin Level': income_binned.ravel()
})
print(df_binned)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import KBinsDiscretizer

bin_edges = discretizer.bin_edges_[0]

fig, (ax1) = plt.subplots(1, 1, figsize=(14, 5))

colors = ['#4C72B0', '#55A868', '#C44E52']
labels = ['Bin 0', 'Bin 1', 'Bin 2']

# -------------------------------------------------------------
# 1. Empirical Density (Histogram + KDE + Rug plot)
# -------------------------------------------------------------
sns.histplot(income.ravel(), bins=10, kde=True, ax=ax1, color='slategray', stat='density', alpha=0.35)
sns.rugplot(income.ravel(), ax=ax1, color='black', height=0.08, linewidth=2)

for i in range(len(bin_edges) - 1):
    ax1.axvspan(bin_edges[i], bin_edges[i+1], color=colors[i], alpha=0.2, 
                label=f"{labels[i]}: [{bin_edges[i]:,.0f}, {bin_edges[i+1]:,.0f}]")
    ax1.axvline(bin_edges[i], color='black', linestyle='--', linewidth=1.2)
ax1.axvline(bin_edges[-1], color='black', linestyle='--', linewidth=1.2)

ax1.set_title('Empirical Density & KDE with Bins', fontweight='bold')
ax1.set_xlabel('Income ($)')
ax1.set_ylabel('Density')
ax1.legend(loc='upper right')
ax1.grid(True, linestyle=':', alpha=0.6)
```

## Feature Scaling

Disparate feature ranges introduce scale dominance in distance calculations (e.g., Euclidean distance $\sqrt{\sum \vert{}x_k\vert{}^2}$) and slow down gradient optimization.   

### Methods

- **Normalization** (Min-Max Scaling): Scales data to a restricted range, typically $[0, 1]$:
   $$x_{\text{norm}, i} = \frac{x_i - x_{\min}}{x_{\max} - x_{\min}}$$
Implemented via `sklearn.preprocessing.MinMaxScaler`.   

- **Standardization** (Z-score Normalization): Centers data to zero mean ($\mu_x = 0$) and unit variance ($\sigma_x = 1$):
   $$x_{\text{std}, i} = \frac{x_i - \mu_x}{\sigma_x}$$
Implemented via `sklearn.preprocessing.StandardScaler`. Standard scaling is less sensitive to extreme outliers than Min-Max scaling.

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

X = np.array([[100.0, 0.001],
              [200.0, 0.050],
              [300.0, 0.005],
              [400.0, 0.100]])

# Min-Max Scaling
min_max_scaler = MinMaxScaler()
X_minmax = min_max_scaler.fit_transform(X)

# Standardization
std_scaler = StandardScaler()
X_std = std_scaler.fit_transform(X)

print("Min-Max Scaled Data:\n", X_minmax)
print("\nStandardized Data:\n", X_std)
```

## Preprocessing Pipelines & ColumnTransformers

Real-world datasets contain heterogeneous feature types (e.g., numerical ratings alongside string locations), requiring separate transformation chains without data leakage.   

### Key Objects

- **Pipeline**: Sequences multiple processing stages into a single composite estimator. All intermediate stages must be `Transformers`; the final stage can be a `Transformer` or `Predictor`.   

- **ColumnTransformer**: Applies designated transformers or pipelines to targeted subsets of feature columns. 

```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression

# 1. Heterogeneous Dataset Construction
raw_df = pd.DataFrame({
    'City': ['London', 'London', 'Paris', 'Sallisaw', np.nan],
    'Expert_Rating': [5.0, 3.0, 4.0, 5.0, 2.0],
    'User_Rating': [4.0, np.nan, 4.0, 3.0, 1.0],
    'Target': [1, 0, 1, 0, 0]
})

X = raw_df.drop(columns=['Target'])
y = raw_df['Target']

print(X)
```

```python
# 2. Define Feature Type Subsets
num_features = ['Expert_Rating', 'User_Rating']
cat_features = ['City']
```

```python
# 3. Sub-Pipelines for Specific Data Types
num_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
# 4. ColumnTransformer Aggregation
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_features),
        ('cat', cat_pipeline, cat_features)
    ],
    remainder='passthrough'
)
```

```python
# 5. Full End-to-End Estimation Pipeline
full_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression())
])
```

```python
# Fit and execute the combined pipeline
full_pipeline.fit(X, y)
predictions = full_pipeline.predict(X)

print("Transformed Feature Matrix Shape:", preprocessor.fit_transform(X).shape)
print("Pipeline Predictions:", predictions)
```
